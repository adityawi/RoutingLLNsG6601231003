```python
# === MMDEW FINAL PDR Parser & Plot ===
import pandas as pd
import matplotlib.pyplot as plt
import re
from pathlib import Path

# 1) Path CSV per skenario (sesuaikan)
file_paths = {
 "P1": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval301.csv",
    "P2": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval302.csv",
    "P3": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval303.csv",
"P4": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval304.csv",
    "P5": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval305.csv",

}

# 2) Regex untuk baris FINAL PDR
#    Menangkap Received, Lost, dan PDR% (angka saja). Fleksibel terhadap spasi.
pdr_pattern = re.compile(
    r"FINAL\s+PDR:\s*Received\s*=\s*(\d+)\s*,\s*Lost\s*=\s*(\d+)\s*,\s*PDR\s*=\s*(\d+)\s*%",
    flags=re.IGNORECASE
)

# 3) Ekstraksi FINAL PDR ke DataFrame
rows = []
for versi, path in file_paths.items():
    if not path or not Path(path).exists():
        print(f"[LEWATI] File tidak ditemukan: {path}")
        continue

    df = pd.read_csv(path)

    # Deteksi nama kolom yang berisi pesan log
    msg_col = "message" if "message" in df.columns else df.columns[-1]
    # (Opsional) deteksi kolom node/id
    node_col = None
    for c in df.columns:
        if c.lower() in ("node", "id", "mote", "source", "src"):
            node_col = c
            break

    for _, r in df.iterrows():
        msg = str(r.get(msg_col, ""))
        m = pdr_pattern.search(msg)
        if m:
            received = int(m.group(1))
            lost     = int(m.group(2))
            pdr_pct  = float(m.group(3))  # sudah persen (0-100)
            rows.append({
                "Versi": versi,
                "Node": r.get(node_col) if node_col in r else None,
                "Received": received,
                "Lost": lost,
                "PDR (%)": pdr_pct
            })

df_pdr = pd.DataFrame(rows)

if df_pdr.empty:
    raise ValueError("Tidak ditemukan baris 'FINAL PDR' pada file-file yang diberikan. Cek path/format log.")

# 4) Rekap: rata-rata PDR per Versi
avg_pdr_per_versi = (
    df_pdr.groupby("Versi", as_index=False)["PDR (%)"].mean().sort_values("Versi")
)

# (Opsional) Rekap per Node per Versi
avg_pdr_per_node = (
    df_pdr.groupby(["Versi", "Node"], as_index=False)["PDR (%)"].mean()
)

print("Rata-rata FINAL PDR per Versi (%):")
display(avg_pdr_per_versi)

# 5) Visualisasi: Bar chart rata-rata FINAL PDR per Versi
plt.figure(figsize=(8, 5))
bars = plt.bar(avg_pdr_per_versi["Versi"], avg_pdr_per_versi["PDR (%)"])

# Tampilkan nilai di atas batang
for b in bars:
    h = b.get_height()
    plt.text(b.get_x() + b.get_width()/2, h, f"{h:.1f}%", ha="center", va="bottom")

plt.xlabel("Jumlah Node")
plt.ylabel("Rata-rata FINAL PDR (%)")
plt.title("Rata-rata FINAL PDR (MMDEW) H4 Interval 10")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 6) Simpan rekap ke CSV (opsional)
out_dir = Path("./pdr_mrhof_output")
out_dir.mkdir(parents=True, exist_ok=True)
avg_pdr_per_versi.to_csv(out_dir / "avg_pdr_per_versi.csv", index=False)
avg_pdr_per_node.to_csv(out_dir / "avg_pdr_per_node.csv", index=False)
print(f"CSV tersimpan di: {out_dir.resolve()}")
```

RSSI

```python
# === MMDEW RSSI Parser & Plot (Mean only, Outlier Handling) ===
import pandas as pd
import matplotlib.pyplot as plt
import re
from pathlib import Path

# 1) Path CSV per skenario (sesuaikan)
file_paths = {
"P1": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/2 April/CSV/MMDEW/Mob5/30/MMDEWmob5Interval301.csv",
    "P2": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/2 April/CSV/MMDEW/Mob5/30/MMDEWmob5Interval302.csv",
    "P3": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/2 April/CSV/MMDEW/Mob5/30/MMDEWmob5Interval303.csv",
"P4": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/2 April/CSV/MMDEW/Mob5/30/MMDEWmob5Interval304.csv",
    "P5": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/2 April/CSV/MMDEW/Mob5/30/MMDEWmob5Interval305.csv",
}

# 2) Regex RSSI=-39dBm (fleksibel spasi, tanda minus, dan desimal)
rssi_pattern = re.compile(r"RSSI\s*=\s*(-?\d+(?:\.\d+)?)\s*dBm", flags=re.IGNORECASE)

# === Parameter outlier ===
# RSSI wajar di WSN biasanya -30 s/d -100 dBm (negatif). Di bawah -120 atau di atas -10 kita anggap anomali.
LOWER_RSSI_DBM  = -120.0
UPPER_RSSI_DBM  = -10.0
USE_IQR = True  # gunakan filter tambahan IQR per versi

# 3) Ekstraksi RSSI mentah
rows = []
for versi, path in file_paths.items():
    if not path or not Path(path).exists():
        print(f"[LEWATI] File tidak ditemukan: {path}")
        continue

    df = pd.read_csv(path)
    msg_col = "message" if "message" in df.columns else df.columns[-1]

    for _, r in df.iterrows():
        msg = str(r.get(msg_col, ""))
        m = rssi_pattern.search(msg)
        if m:
            rssi_dbm = float(m.group(1))
            rows.append({"Versi": versi, "RSSI (dBm)": rssi_dbm})

df_rssi_raw = pd.DataFrame(rows)
if df_rssi_raw.empty:
    raise ValueError("Tidak ditemukan log 'RSSI=...dBm' pada file-file yang diberikan.")

# 4) Buang outlier (threshold + IQR per versi)
# Threshold fisik
df_rssi = df_rssi_raw[(df_rssi_raw["RSSI (dBm)"] >= LOWER_RSSI_DBM) &
                      (df_rssi_raw["RSSI (dBm)"] <= UPPER_RSSI_DBM)].copy()

removed_thresh = len(df_rssi_raw) - len(df_rssi)

# IQR per versi (opsional)
removed_iqr = 0
if USE_IQR and not df_rssi.empty:
    cleaned = []
    for versi, g in df_rssi.groupby("Versi"):
        Q1 = g["RSSI (dBm)"].quantile(0.25)
        Q3 = g["RSSI (dBm)"].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        cg = g[(g["RSSI (dBm)"] >= lower) & (g["RSSI (dBm)"] <= upper)]
        removed_iqr += len(g) - len(cg)
        cleaned.append(cg)
    df_rssi = pd.concat(cleaned, ignore_index=True)

print(f"[INFO] Sampel awal: {len(df_rssi_raw)} | Dibuang (threshold): {removed_thresh} | Dibuang (IQR): {removed_iqr}")
if df_rssi.empty:
    raise ValueError("Semua data RSSI terfilter. Sesuaikan batas threshold/IQR.")

# 5) Rata-rata RSSI per Versi (lebih tinggi = lebih dekat ke 0 = sinyal lebih kuat)
avg_rssi_per_versi = (
    df_rssi.groupby("Versi", as_index=False)["RSSI (dBm)"].mean().sort_values("Versi")
)

print("Rata-rata RSSI per Versi (dBm):")
display(avg_rssi_per_versi)

# 6) Visualisasi: Bar chart mean RSSI
plt.figure(figsize=(8, 5))
bars = plt.bar(avg_rssi_per_versi["Versi"], avg_rssi_per_versi["RSSI (dBm)"])

for b in bars:
    h = b.get_height()
    plt.text(b.get_x() + b.get_width()/2, h, f"{h:.1f}", ha="center", va="bottom")

plt.xlabel("Jumlah Node")
plt.ylabel("Rata-rata RSSI (dBm)")
plt.title("Rata-rata RSSI (MMDEW) — N 20 Interval 20")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 7) Simpan CSV (opsional)
from pathlib import Path
out_dir = Path("./rssi_mmdew_output")
out_dir.mkdir(parents=True, exist_ok=True)
avg_rssi_per_versi.to_csv(out_dir / "rssi_summary_mean.csv", index=False)
print(f"CSV tersimpan di: {out_dir.resolve()}")
```

```python
# === MRHOF Delay Parser & Plot (Mean only, Outlier Handling) ===
import pandas as pd
import matplotlib.pyplot as plt
import re
from pathlib import Path

# 1) Path CSV per skenario (sesuaikan)
file_paths = {
"P1": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval301.csv",
    "P2": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval302.csv",
    "P3": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval303.csv",
"P4": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval304.csv",
    "P5": "/content/drive/MyDrive/LOG CONTIKI-NG/DTW-RL/3 Juli/CSV/Mobility5/MMDEW/30/MMDEWmob5Interval305.csv",

}

# 2) Regex Delay=...ms
delay_pattern = re.compile(r"Delay\s*=\s*(\d+)\s*ms", flags=re.IGNORECASE)

# === Parameter outlier ===
HARD_THRESHOLD_MS = 10_000  # buang Delay > 10 detik
USE_IQR = True              # filter outlier dengan IQR

# 3) Ekstraksi Delay
rows = []
for versi, path in file_paths.items():
    if not path or not Path(path).exists():
        print(f"[LEWATI] File tidak ditemukan: {path}")
        continue

    df = pd.read_csv(path)
    msg_col = "message" if "message" in df.columns else df.columns[-1]

    for _, r in df.iterrows():
        msg = str(r.get(msg_col, ""))
        m = delay_pattern.search(msg)
        if m:
            delay_ms = int(m.group(1))
            rows.append({"Versi": versi, "Delay (ms)": delay_ms})

df_delay_raw = pd.DataFrame(rows)
if df_delay_raw.empty:
    raise ValueError("Tidak ditemukan log 'Delay=...ms'.")

# 4) Buang outlier
df_delay = df_delay_raw.copy()

if HARD_THRESHOLD_MS is not None:
    df_delay = df_delay[df_delay["Delay (ms)"] <= HARD_THRESHOLD_MS]

if USE_IQR and not df_delay.empty:
    cleaned_list = []
    for versi, g in df_delay.groupby("Versi"):
        Q1 = g["Delay (ms)"].quantile(0.25)
        Q3 = g["Delay (ms)"].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        cleaned = g[(g["Delay (ms)"] >= lower) & (g["Delay (ms)"] <= upper)]
        cleaned_list.append(cleaned)
    df_delay = pd.concat(cleaned_list, ignore_index=True)

# 5) Hitung rata-rata Delay per Versi
avg_delay_per_versi = (
    df_delay.groupby("Versi", as_index=False)["Delay (ms)"].mean().sort_values("Versi")
)

print("Rata-rata Delay per Versi (ms):")
display(avg_delay_per_versi)

# 6) Visualisasi: Bar chart mean Delay
plt.figure(figsize=(8, 5))
bars = plt.bar(avg_delay_per_versi["Versi"], avg_delay_per_versi["Delay (ms)"], color="skyblue")

for b in bars:
    h = b.get_height()
    plt.text(b.get_x() + b.get_width()/2, h, f"{h:.1f}", ha="center", va="bottom")

plt.xlabel("Jumlah Node")
plt.ylabel("Rata-rata Delay (ms)")
plt.title("Rata-rata Delay (MMDEW) — N20 Interval 20")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 7) Simpan CSV (opsional)
out_dir = Path("./delay_mrhof_output")
out_dir.mkdir(parents=True, exist_ok=True)
avg_delay_per_versi.to_csv(out_dir / "delay_summary_mean.csv", index=False)
print(f"CSV tersimpan di: {out_dir.resolve()}")
```

```python
import re
import pandas as pd

# -------------------------------------------------------------------
# 1. Load log Contiki-NG
# -------------------------------------------------------------------
LOG_PATH = "/content/drive/MyDrive/LOG CONTIKI-NG/ujicoba/DTW-RL.csv"   # sesuaikan dengan lokasi file Anda

df = pd.read_csv(LOG_PATH)

# Pastikan kolom sesuai: time, node, message
print(df.columns)

# -------------------------------------------------------------------
# 2. Ekstrak konfigurasi awal: "DTW-RL tau-start=..."
# -------------------------------------------------------------------
def extract_tau_start(df):
    """
    Mengambil informasi tau-start (τ awal) DTW-RL per node.
    Output: DataFrame dengan kolom:
        - time      : waktu simulasi saat tau-start dicetak
        - node      : id node
        - tau_start : nilai τ awal (detik, float)
    """
    # Filter baris yang mengandung teks "DTW-RL tau-start"
    mask_tau = df["message"].str.contains("DTW-RL tau-start", na=False)
    df_tau = df[mask_tau].copy()

    # Ekstrak angka setelah "tau-start=" sampai "s"
    df_tau["tau_start"] = (
        df_tau["message"]
        .str.extract(r"tau-start=([0-9.]+)s", expand=False)
        .astype(float)
    )

    # Rapikan nama kolom
    df_tau = df_tau[["time", "node", "tau_start"]].reset_index(drop=True)
    return df_tau

# -------------------------------------------------------------------
# 3. Ekstrak kontrol DTW-RL per window: "DTW-RL SUM i=...s | winQ=x/y"
# -------------------------------------------------------------------
def extract_dtw_windows(df):
    """
    Mengambil ringkasan window DTW-RL dari log "DTW-RL SUM".
    Output: DataFrame dengan kolom:
        - time         : waktu simulasi saat window selesai
        - node         : id node
        - window_len   : durasi window (τ_i) dalam detik
        - winQ_valid   : jumlah sample valid (lolos evidence gate)
        - winQ_total   : total sample dalam window
    """
    # Filter baris yang mengandung "DTW-RL SUM"
    mask_sum = df["message"].str.contains("DTW-RL SUM", na=False)
    df_sum = df[mask_sum].copy()

    # Kadang huruf 'i' di-log bisa korup (Ï), jadi regex dibuat toleran
    # Pola: "i=24.69s" atau "Ï=24.69s"
    df_sum["window_len"] = (
        df_sum["message"]
        .str.extract(r"[iÏ]=([0-9.]+)s", expand=False)
        .astype(float)
    )

    # Ekstrak winQ=x/y
    winq = df_sum["message"].str.extract(r"winQ=(\d+)/(\d+)", expand=True)
    winq.columns = ["winQ_valid", "winQ_total"]
    df_sum["winQ_valid"] = winq["winQ_valid"].astype("Int64")
    df_sum["winQ_total"] = winq["winQ_total"].astype("Int64")

    # Rapikan
    df_sum = df_sum[["time", "node", "window_len", "winQ_valid", "winQ_total"]]
    df_sum = df_sum.reset_index(drop=True)
    return df_sum

# -------------------------------------------------------------------
# 4. Jalankan ekstraksi & simpan ke file CSV terpisah (opsional)
# -------------------------------------------------------------------
df_tau_start = extract_tau_start(df)
df_windows   = extract_dtw_windows(df)

print("=== Tau-start DTW-RL ===")
print(df_tau_start.head())

print("\n=== DTW-RL Windows Summary ===")
print(df_windows.head())

# Simpan ke file jika ingin dianalisis di luar Python (Excel, R, dsb.)
df_tau_start.to_csv("dtwrl_tau_start.csv", index=False)
df_windows.to_csv("dtwrl_windows.csv", index=False)
```

```python
import re
import pandas as pd

# -------------------------------------------------------------
# 0. Load log (kalau df sudah ada, bagian ini boleh di-skip)
# -------------------------------------------------------------
LOG_PATH = "/content/drive/MyDrive/LOG CONTIKI-NG/ujicoba/DTW-RL.csv"
df = pd.read_csv(LOG_PATH)   # pastikan kolom: time, node, message

# =============================================================
# 1. EKSTRAK RL-SELECT (state, Q1, Q2, selected parent)
# =============================================================

def extract_rl_select(df):
    """
    Ekstrak baris log:
        [INFO: RPL-MMDEW ] RL-SELECT s=2 | Q1=-2.000 Q2=-2.000 | sel=p2

    Output: DataFrame dengan kolom:
        - time
        - node
        - state
        - Q1, Q2, ... (jika ada Q3, Q4 akan ikut terbuat otomatis)
        - selected_parent
    """
    # Filter hanya baris yang mengandung "RL-SELECT"
    mask = df["message"].str.contains("RL-SELECT", na=False)
    sub = df[mask].copy()

    # --- Ekstrak state ---
    sub["state"] = (
        sub["message"]
        .str.extract(r"RL-SELECT\s+s=(\d+)", expand=False)
        .astype("Int64")
    )

    # --- Ekstrak selected parent (sel=p1/p2/...) ---
    sub["selected_parent"] = sub["message"].str.extract(r"sel=(p\d+)", expand=False)

    # --- Ekstrak semua pasangan Qk=nilai secara generik ---
    # Contoh: "Q1=-2.000 Q2=0.050"
    def parse_Qs(msg):
        # hasil: dict { 'Q1': -2.0, 'Q2': 0.05, ... }
        pairs = re.findall(r"(Q\d+)=(-?\d+\.\d+)", msg)
        return {name: float(val) for name, val in pairs}

    q_dicts = sub["message"].apply(parse_Qs)
    q_df = pd.DataFrame(q_dicts.tolist())  # kolom Q1, Q2, ... otomatis

    # Gabungkan semua informasi
    sub = pd.concat([sub[["time", "node", "state", "selected_parent"]], q_df], axis=1)

    # Reset index
    sub = sub.reset_index(drop=True)
    return sub

# =============================================================
# 2. EKSTRAK DTW-RL ASYNC penalize (event penalti RL)
# =============================================================

def extract_rl_penalize(df):
    """
    Ekstrak baris log:
        [INFO: RPL-MMDEW ] DTW-RL ASYNC penalize c1:0c Q0=0.050

    Output: DataFrame dengan kolom:
        - time
        - node
        - counter_raw   : string counter (mis. 'c1:0c')
        - penalized_id  : jika pola 'c<id>:' -> id parent/slot yang kena penalti
        - Q0            : nilai Q0 yang tercetak pada log
    """
    # Filter baris yang mengandung "DTW-RL ASYNC penalize"
    mask = df["message"].str.contains("DTW-RL ASYNC penalize", na=False)
    sub = df[mask].copy()

    # --- Ekstrak counter string, mis. 'c1:0c' ---
    sub["counter_raw"] = sub["message"].str.extract(r"penalize\s+(\S+)", expand=False)

    # --- Optional: ekstrak id dari pola 'c<id>:' kalau memang merepresentasikan parent ---
    # contoh: 'c1:0c' -> penalized_id = 1
    def get_penalized_id(counter):
        if isinstance(counter, str):
            m = re.search(r"c(\d+):", counter)
            if m:
                return int(m.group(1))
        return pd.NA

    sub["penalized_id"] = sub["counter_raw"].apply(get_penalized_id).astype("Int64")

    # --- Ekstrak Q0 ---
    sub["Q0"] = (
        sub["message"]
        .str.extract(r"Q0=(-?\d+\.\d+)", expand=False)
        .astype(float)
    )

    # Pilih kolom penting
    sub = sub[["time", "node", "counter_raw", "penalized_id", "Q0"]]
    sub = sub.reset_index(drop=True)
    return sub

# =============================================================
# 3. JALANKAN EKSTRAKSI & SIMPAN
# =============================================================

df_rl_select   = extract_rl_select(df)
df_rl_penalize = extract_rl_penalize(df)

print("=== RL-SELECT decisions ===")
print(df_rl_select.head())

print("\n=== RL penalize events ===")
print(df_rl_penalize.head())

# Simpan ke file CSV (opsional)
df_rl_select.to_csv("dtwrl_rl_select.csv", index=False)
df_rl_penalize.to_csv("dtwrl_rl_penalize.csv", index=False)
```

```python
# ============================================================
# AWT - DTW-RL LOG ANALYSIS (Node 7)
# Fokus:
# 1) Q-table per window (snapshot)
# 2) Overlay Q(s,a) vs τ (shrink τ → update Q?)
# ============================================================

import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# 0) PATH LOG (ubah sesuai lokasi)
# -----------------------------
LOG_PATH = "/content/drive/MyDrive/LOG CONTIKI-NG/ujicoba/DTW-RL.csv"

# -----------------------------
# 1) Load log mentah
# -----------------------------
df = pd.read_csv(LOG_PATH)
required_cols = {"time", "node", "message"}
if not required_cols.issubset(df.columns):
    raise ValueError(f"Kolom wajib tidak lengkap. Dapat: {df.columns.tolist()}")

# -----------------------------
# 2) Helper: time mm:ss.xxx -> detik
# -----------------------------
def time_to_seconds(t):
    if not isinstance(t, str) or ":" not in t:
        return np.nan
    m, rest = t.split(":")
    try:
        return int(m) * 60 + float(rest)
    except:
        return np.nan

df["t_sec"] = df["time"].apply(time_to_seconds)
df = df.dropna(subset=["t_sec"]).copy()
df = df.sort_values(["node", "t_sec"]).reset_index(drop=True)

# ============================================================
# 3) EKSTRAK WINDOW (DTW-RL SUM) -> tau per window
# ============================================================
def extract_dtw_windows(df):
    mask = df["message"].str.contains("DTW-RL SUM", na=False)
    sub = df.loc[mask, ["time", "node", "t_sec", "message"]].copy()

    # i=24.69s atau Ï=24.69s (handle encoding)
    sub["tau"] = sub["message"].str.extract(r"[iÏ]=([0-9.]+)s", expand=False).astype(float)

    winq = sub["message"].str.extract(r"winQ=(\d+)/(\d+)", expand=True)
    sub["winQ_valid"] = winq[0].astype("Int64")
    sub["winQ_total"] = winq[1].astype("Int64")

    # window marker: setiap baris SUM = akhir sebuah window
    sub = sub.drop(columns=["message"]).reset_index(drop=True)
    return sub

df_win = extract_dtw_windows(df)

# ============================================================
# 4) EKSTRAK RL-SELECT -> Q(s,a) pada waktu keputusan
# ============================================================
def extract_rl_select(df):
    mask = df["message"].str.contains("RL-SELECT", na=False)
    sub = df.loc[mask, ["time", "node", "t_sec", "message"]].copy()

    sub["state"] = sub["message"].str.extract(r"RL-SELECT\s+s=(\d+)", expand=False).astype("Int64")
    sub["selected_parent"] = sub["message"].str.extract(r"sel=(p\d+)", expand=False)

    def parse_qpairs(msg):
        # return list of (action, qvalue) e.g. [('p1',0.55), ('p2',-2.0)]
        pairs = re.findall(r"(Q\d+)=(-?\d+\.\d+)", msg)
        out = []
        for qname, qval in pairs:
            action = qname.replace("Q", "p")  # Q1 -> p1
            out.append((action, float(qval)))
        return out

    rows = []
    for _, r in sub.iterrows():
        pairs = parse_qpairs(r["message"])
        for action, qval in pairs:
            rows.append({
                "time": r["time"],
                "node": r["node"],
                "t_sec": r["t_sec"],
                "state": r["state"],
                "action": action,
                "Q": qval,
                "selected_parent": r["selected_parent"]
            })

    out = pd.DataFrame(rows)
    return out.sort_values(["node", "t_sec"]).reset_index(drop=True)

df_q = extract_rl_select(df)

# ============================================================
# 5) Fokus Node 7
# ============================================================
NODE_ID = 7

win7 = df_win[df_win["node"] == NODE_ID].copy().sort_values("t_sec").reset_index(drop=True)
q7   = df_q[df_q["node"] == NODE_ID].copy().sort_values("t_sec").reset_index(drop=True)

if len(win7) == 0:
    raise ValueError("Tidak ada data DTW-RL SUM untuk Node 7 (win7 kosong).")
if len(q7) == 0:
    raise ValueError("Tidak ada data RL-SELECT untuk Node 7 (q7 kosong).")

# Tambahkan index window (0..n-1) berdasarkan urutan DTW-RL SUM
win7["window_id"] = np.arange(len(win7), dtype=int)

# Definisikan rentang window: (end_prev, end_now]
# start window_0 = 0 (atau tau-start). Kita pakai 0 agar aman.
win7["window_start"] = win7["t_sec"].shift(1, fill_value=0.0)
win7["window_end"]   = win7["t_sec"]

# ============================================================
# 6) Assign setiap Q log ke window terdekat sebelumnya (merge_asof)
#    -> Q event mendapat window_id dan tau window saat itu
# ============================================================
# kita join berdasarkan "t_sec" (waktu Q-event) ke "window_end" terakhir <= t_sec
tmp_win = win7[["window_id", "window_end", "tau", "winQ_valid", "winQ_total"]].copy().rename(columns={"window_end":"t_sec"})
q7_tag = pd.merge_asof(
    q7.sort_values("t_sec"),
    tmp_win.sort_values("t_sec"),
    on="t_sec",
    direction="backward"
)

# Jika ada event Q sebelum window pertama selesai, window_id akan NaN -> set ke 0 (opsional)
q7_tag["window_id"] = q7_tag["window_id"].fillna(0).astype(int)

# ============================================================
# 7) Bangun Q-table per window (snapshot)
#    Definisi snapshot: Q TERAKHIR (last seen) untuk setiap (state, action) di dalam window tsb
# ============================================================
# Kita ambil "baris terakhir" per (window_id, state, action)
q_last = (
    q7_tag.sort_values("t_sec")
          .groupby(["window_id", "state", "action"], as_index=False)
          .tail(1)
          .sort_values(["window_id", "state", "action"])
)

# Pivot agar menjadi Q-table per window: index=(state, action) atau state sebagai index, action sebagai columns
# Untuk kemudahan, kita buat table per window: (state x action)
def qtable_for_window(wid):
    sub = q_last[q_last["window_id"] == wid]
    if len(sub) == 0:
        return None
    return sub.pivot(index="state", columns="action", values="Q").sort_index()

# Contoh tampilkan 5 window pertama
print("=== Window summary (Node 7) ===")
print(win7.head(10))

print("\n=== Contoh Q-table snapshot Node 7 (5 window pertama yang punya data) ===")
shown = 0
for wid in sorted(q_last["window_id"].unique()):
    qt = qtable_for_window(wid)
    if qt is None:
        continue
    print(f"\n--- window_id={wid} | τ={win7.loc[win7['window_id']==wid,'tau'].values[0]:.2f}s ---")
    print(qt)
    shown += 1
    if shown >= 5:
        break

# ============================================================
# 8) VISUALISASI A: Overlay Q(s,a) vs τ
#    - τ sebagai garis (left axis)
#    - Q-event sebagai titik/garis (right axis)
# ============================================================

# 8.1 Plot τ evolution (window_end vs tau)
plt.figure(figsize=(10,4))
plt.plot(win7["window_end"], win7["tau"], marker="o", linewidth=2)
plt.xlabel("Time (s)")
plt.ylabel("τ (window length, s)")
plt.title("Node 7 — Evolusi τ per window")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

# 8.2 Overlay τ dan Q events (dual y-axis)
fig, ax1 = plt.subplots(figsize=(11,4))

# τ line
ax1.plot(win7["window_end"], win7["tau"], marker="o", linewidth=2)
ax1.set_xlabel("Time (s)")
ax1.set_ylabel("τ (s)")
ax1.grid(True, linestyle="--", alpha=0.6)

# Deteksi shrink window: tau_t < tau_{t-1}
shrink_idx = win7.index[win7["tau"].diff() < 0].tolist()
ax1.scatter(win7.loc[shrink_idx, "window_end"], win7.loc[shrink_idx, "tau"], marker="x", s=80)

# Q points on ax2
ax2 = ax1.twinx()
# plot Q per action (p1, p2, dst.)
for action, g in q7_tag.groupby("action"):
    ax2.plot(g["t_sec"], g["Q"], marker=".", linestyle="-", alpha=0.9, label=f"Q {action}")

ax2.set_ylabel("Q-value")

# Legend gabungan
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines1 + lines2, labels1 + labels2, loc="best")

plt.title("Node 7 — Overlay τ vs Q(s,a) (X = shrink τ)")
plt.tight_layout()
plt.show()

# ============================================================
# 9) UJI SEDERHANA: Apakah shrink τ diikuti peningkatan update Q?
#    Kita hitung jumlah Q-events di setiap window, lalu bandingkan window shrink vs non-shrink
# ============================================================

# Hitung jumlah Q-events per window
q_count = q7_tag.groupby("window_id").size().rename("q_events").reset_index()
win7_eval = win7.merge(q_count, on="window_id", how="left")
win7_eval["q_events"] = win7_eval["q_events"].fillna(0).astype(int)

# label shrink untuk window t (dibanding window sebelumnya)
win7_eval["is_shrink"] = win7_eval["tau"].diff() < 0
win7_eval["is_shrink"] = win7_eval["is_shrink"].fillna(False)

print("\n=== Ringkasan Q-events per window (Node 7) ===")
print(win7_eval[["window_id", "window_end", "tau", "is_shrink", "q_events", "winQ_valid", "winQ_total"]].head(15))

print("\n=== Rata-rata Q-events: shrink vs non-shrink ===")
print(win7_eval.groupby("is_shrink")["q_events"].mean())

# Plot q_events per window dengan highlight shrink
plt.figure(figsize=(10,4))
plt.bar(win7_eval["window_id"], win7_eval["q_events"])
plt.scatter(win7_eval.loc[win7_eval["is_shrink"], "window_id"],
            win7_eval.loc[win7_eval["is_shrink"], "q_events"],
            marker="x", s=80)
plt.xlabel("window_id")
plt.ylabel("#Q events in window")
plt.title("Node 7 — #Q events per window (X = shrink τ)")
plt.grid(True, axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

# ============================================================
# 10) (Opsional) Simpan output agar gampang dipakai (Excel/paper)
# ============================================================
# q7_tag.to_csv("/content/node7_q_events_tagged.csv", index=False)
# win7_eval.to_csv("/content/node7_windows_eval.csv", index=False)
```
